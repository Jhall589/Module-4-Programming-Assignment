```python
#16.4
import sqlite3

# Create/connect to database
conn = sqlite3.connect('books.db')

# Create cursor
cur = conn.cursor()

# Create books table
cur.execute('''
    CREATE TABLE books (
        title TEXT,
        author TEXT,
        year INTEGER
    )
''')

# Save changes
conn.commit()

# Close connection
conn.close()

print("books.db and books table created.")
```

```python
import sqlite3
import csv

conn = sqlite3.connect('books.db')
cur = conn.cursor()

with open('books2.csv', 'r', encoding='utf-8-sig') as file:
    reader = csv.reader(file)

    next(reader)

    for row in reader:
        cur.execute(
            "INSERT INTO books (title, author, year) VALUES (?, ?, ?)",
            (row[0], row[1], int(row[2]))
        )

conn.commit()
conn.close()

print("Data inserted successfully.")
```
